# Cálculo da Receita, Lucro por Unidade, Lucro Final

```python
df_vendas_geek_store["Receita"] = df_vendas_geek_store['Preco_Venda'] * df_vendas_geek_store['Quantidade_Vendida']
df_vendas_geek_store["Lucro por Unidade"] = df_vendas_geek_store['Preco_Venda'] - df_vendas_geek_store['Custo_Fabricante']
df_vendas_geek_store["Lucro Final"] = df_vendas_geek_store['Lucro por Unidade'] * df_vendas_geek_store['Quantidade_Vendida']

display(df_vendas_geek_store[['Produto','Receita','Lucro por Unidade','Lucro Final']])
```

# Filtro de componentes de PC

```python
df_filtrado = df_vendas_geek_store[df_vendas_geek_store['Categoria'] == 'Hardware']
display(df_filtrado[['Produto', 'Categoria']])
```
